```python
def two_sum_sorted(nums, target):

    # STEP 1:
    # We use two pointers to search for the two numbers.
    #
    # 'left' starts at the beginning of the list.
    # 'right' starts at the end of the list.
    #
    # Example:
    # nums = [2, 7, 11, 15]
    #
    # left  -> 2
    # right -> 15
    #
    # Because the array is sorted, we can decide
    # which pointer to move based on the current sum.
    left = 0
    right = len(nums) - 1

    # STEP 2:
    # Continue searching while the two pointers
    # have not crossed each other.
    #
    # left < right means:
    # There are still at least two different
    # elements available to check.
    while left < right:

        # STEP 3:
        # Calculate the sum of the two numbers
        # currently pointed to by left and right.
        #
        # Example:
        # nums = [2, 7, 11, 15]
        #
        # left = 0
        # right = 3
        #
        # nums[left]  = 2
        # nums[right] = 15
        #
        # current_sum = 2 + 15 = 17
        current_sum = nums[left] + nums[right]

        # STEP 4:
        # Check whether the current sum is exactly
        # equal to the target.
        #
        # If it is, we have found the two numbers
        # we are looking for.
        #
        # Example:
        # target = 17
        #
        # 2 + 15 = 17
        #
        # Therefore, return their indices.
        if current_sum == target:
            return [left, right]

        # STEP 5:
        # If the current sum is smaller than the target,
        # we need a BIGGER sum.
        #
        # Because nums is sorted in ascending order,
        # moving 'left' to the right gives us
        # a larger number.
        #
        # Example:
        # nums = [2, 7, 11, 15]
        # target = 18
        #
        # Current:
        # 2 + 15 = 17
        #
        # 17 is too small.
        #
        # Move left forward:
        # left -> 7
        #
        # Now:
        # 7 + 15 = 22
        elif current_sum < target:
            left += 1

        # STEP 6:
        # If the current sum is greater than the target,
        # we need a SMALLER sum.
        #
        # Because nums is sorted,
        # moving 'right' to the left gives us
        # a smaller number.
        #
        # Example:
        # nums = [2, 7, 11, 15]
        # target = 17
        #
        # Suppose:
        # 7 + 15 = 22
        #
        # 22 is too large.
        #
        # Move right backward:
        # right -> 11
        #
        # Now:
        # 7 + 11 = 18
        else:
            right -= 1

    # STEP 7:
    # If the while loop finishes, it means
    # we checked all possible pointer combinations
    # and did not find two numbers whose sum
    # equals the target.
    #
    # Returning None means:
    # "No valid pair was found."
    return None
```

```python
def two_sum_sorted(nums, target):

    left = 0
    right = len(nums) - 1

    while left < right:
        current_sum = nums[left] + nums[right]

        if current_sum == target:
            return [left, right]
        elif current_sum < target:
            left += 1
        else:
            right -= 1

    return None
```

```python
def move_zeroes(nums):

    # STEP 1:
    # Create a pointer called 'insert_pos'.
    #
    # This pointer tells us where the next
    # non-zero number should be placed.
    #
    # Initially, we start at index 0 because
    # the first non-zero number should go there.
    #
    # Example:
    # nums = [0, 1, 0, 3, 12]
    #
    # insert_pos = 0
    #
    # This means:
    # "The next non-zero number should be
    # placed at index 0."
    insert_pos = 0

    # STEP 2:
    # Go through every element in the list
    # from left to right.
    #
    # 'i' represents the index of the current
    # element we are examining.
    #
    # Example:
    # i = 0 -> nums[0] = 0
    # i = 1 -> nums[1] = 1
    # i = 2 -> nums[2] = 0
    # i = 3 -> nums[3] = 3
    # i = 4 -> nums[4] = 12
    for i in range(len(nums)):

        # STEP 3:
        # Check whether the current number
        # is NOT zero.
        #
        # We only care about non-zero numbers
        # because our goal is to move all of them
        # toward the beginning of the list.
        #
        # Zeroes will automatically end up
        # toward the end.
        if nums[i] != 0:

            # STEP 4:
            # Put the current non-zero number
            # at the 'insert_pos' position.
            #
            # This keeps all non-zero numbers
            # together at the beginning.
            #
            # Example:
            # nums = [0, 1, 0, 3, 12]
            #
            # When i = 1:
            # nums[i] = 1
            # insert_pos = 0
            #
            # Move 1 to index 0:
            # [1, 1, 0, 3, 12]
            #
            # The original 1 at index 1 is temporarily
            # duplicated. We will replace it with 0
            # later.
            nums[insert_pos] = nums[i]

            # STEP 5:
            # Move 'insert_pos' forward by one.
            #
            # We have successfully placed one
            # non-zero number.
            #
            # Therefore, the next non-zero number
            # should go into the next position.
            insert_pos += 1

    # STEP 6:
    # At this point, all non-zero numbers have been
    # moved to the beginning of the list.
    #
    # The remaining positions from 'insert_pos'
    # to the end of the list should contain zeroes.
    #
    # Example:
    # Original:
    # [0, 1, 0, 3, 12]
    #
    # After collecting non-zero numbers:
    # [1, 3, 12, 3, 12]
    #
    # insert_pos = 3
    #
    # The first 3 positions contain:
    # [1, 3, 12]
    #
    # Now we need to fill positions 3 and 4
    # with zeroes.
    for i in range(insert_pos, len(nums)):

        # STEP 7:
        # Replace every remaining position
        # with zero.
        #
        # Final result:
        # [1, 3, 12, 0, 0]
        nums[i] = 0
```

```python
def move_zeroes(nums):
    insert_pos = 0

    for i in range(len(nums)):
        
        if nums[i] != 0:
            nums[insert_pos] = nums[i]
            insert_pos += 1

    for i in range(insert_pos, len(nums)):
        nums[i] = 0
```

```python
def remove_duplicates(nums):

    # STEP 1:
    # If the array is empty, there are no unique elements.
    #
    # Example:
    # nums = []
    #
    # Answer = 0
    if not nums:
        return 0

    # STEP 2:
    # 'write' points to the position of the
    # LAST UNIQUE ELEMENT we have found.
    #
    # The first element is always unique,
    # so initially:
    #
    # nums = [1, 1, 2, 2, 3]
    #         ^
    #       write
    #
    # write = 0
    write = 0

    # STEP 3:
    # 'read' scans the array from left to right.
    #
    # We start from index 1 because index 0
    # is already considered unique.
    for read in range(1, len(nums)):

        # STEP 4:
        # Compare the current element with
        # the LAST UNIQUE element.
        #
        # nums[read]     -> element we are checking
        # nums[write]    -> last unique element
        #
        # If they are the same:
        #     duplicate → do nothing
        #
        # If they are different:
        #     new unique element found
        if nums[read] != nums[write]:

            # STEP 5:
            # Move 'write' forward because we found
            # another unique element.
            write += 1

            # STEP 6:
            # Copy the new unique element into
            # the 'write' position.
            #
            # Example:
            # nums = [1, 1, 2, 2, 3]
            #
            # We find 2.
            #
            # Before:
            # [1, 1, 2, 2, 3]
            #  ^
            # write = 0
            #
            # Move write:
            # write = 1
            #
            # Copy 2:
            # [1, 2, 2, 2, 3]
            nums[write] = nums[read]

    # STEP 7:
    # 'write' is the INDEX of the last unique element.
    #
    # Therefore, the number of unique elements
    # is write + 1.
    #
    # Example:
    # [1, 2, 3, 2, 3]
    #       ^
    #    write = 2
    #
    # Number of unique elements = 2 + 1 = 3
    #
    # If we want the actual unique elements:
    # nums[0:write + 1] → [1, 2, 3]
    return nums[0:write + 1]
```

```python
def remove_duplicates(nums):
    if not nums:
        return 0

    write = 0

    for read in range(1, len(nums)):
        if nums[read] != nums[write]:
            write += 1
            nums[write] = nums[read]

    return write + 1
```

```python
def max_sum_subarray(nums, k):

    # STEP 1:
    # We need to find the maximum sum of a
    # CONTIGUOUS subarray of size k.
    #
    # Example:
    # nums = [2, 1, 5, 1, 3, 2]
    # k = 3
    #
    # We need to check windows of exactly 3 elements:
    #
    # [2, 1, 5]
    # [1, 5, 1]
    # [5, 1, 3]
    # [1, 3, 2]
    #
    # Instead of calculating every window's sum
    # from scratch, we use the SLIDING WINDOW technique.

    # STEP 2:
    # Handle invalid cases.
    #
    # If the array is empty, or k is 0 or negative,
    # there is no valid subarray to calculate.
    #
    # Also, if k is larger than the array length,
    # we cannot create a window of size k.
    if not nums or k <= 0 or k > len(nums):
        return None

    # STEP 3:
    # Calculate the sum of the FIRST window.
    #
    # Example:
    # nums = [2, 1, 5, 1, 3, 2]
    # k = 3
    #
    # First window:
    # [2, 1, 5]
    #
    # window_sum = 2 + 1 + 5
    #             = 8
    window_sum = sum(nums[:k])

    # STEP 4:
    # At the beginning, the first window is
    # the largest sum we have seen so far.
    #
    # Later, we will compare every new window
    # against this value.
    max_sum = window_sum

    # STEP 5:
    # Start sliding the window.
    #
    # 'right' represents the NEW element
    # entering the window.
    #
    # We start from index k because indices
    # 0 through k-1 were already included
    # in the first window.
    for right in range(k, len(nums)):

        # STEP 6:
        # Add the new element entering the window.
        #
        # Example:
        #
        # Current window:
        # [2, 1, 5]
        #
        # Next element:
        # 1
        #
        # Add 1:
        # 8 + 1 = 9
        window_sum += nums[right]

        # STEP 7:
        # Remove the element that is leaving
        # the window.
        #
        # When the new 1 enters:
        #
        # [2, 1, 5] → [1, 5, 1]
        #
        # The first 2 is no longer part of
        # the current window.
        #
        # So:
        # 9 - 2 = 7
        #
        # IMPORTANT:
        # We add the new element first,
        # then remove the old element.
        window_sum -= nums[right - k]

        # STEP 8:
        # Compare the current window's sum
        # with the maximum sum found so far.
        #
        # If the current window is bigger,
        # update max_sum.
        #
        # Example:
        # Current sum = 9
        # max_sum = 8
        #
        # 9 > 8
        # Therefore:
        # max_sum = 9
        max_sum = max(max_sum, window_sum)

    # STEP 9:
    # After sliding the window across the entire
    # array, max_sum contains the largest sum
    # of any contiguous subarray of size k.
    #
    # Example:
    #
    # nums = [2, 1, 5, 1, 3, 2]
    # k = 3
    #
    # Maximum window:
    # [5, 1, 3]
    #
    # Maximum sum:
    # 9
    return max_sum
```

```python
def max_sum_subarray(nums, k):
    if not nums or k <= 0 or k > len(nums):
        return None

    window_sum = sum(nums[:k])
    max_sum = window_sum

    for right in range(k, len(nums)):
        window_sum += nums[right]          # add new element
        window_sum -= nums[right - k]      # remove old element

        max_sum = max(max_sum, window_sum)

    return max_sum
```

```python
def max_subarray(nums):

    # STEP 1:
    # We start with the first element.
    #
    # 'current_sum' means:
    # Maximum sum of a subarray that MUST end
    # at the current position.
    #
    # 'max_sum' means:
    # Maximum subarray sum found anywhere so far.
    #
    # Example:
    # nums = [-2, 1, -3, 4, -1, 2, 1]
    #
    # Initially:
    # current_sum = -2
    # max_sum = -2
    current_sum = nums[0]
    max_sum = nums[0]

    # STEP 2:
    # Start from the second element because
    # the first element was already used above.
    for i in range(1, len(nums)):

        # STEP 3:
        # We have TWO choices:
        #
        # 1. Start a NEW subarray from nums[i]
        #
        # 2. Add nums[i] to the existing subarray
        #
        # If current_sum is hurting us, we start fresh.
        #
        # Example:
        #
        # current_sum = -5
        # nums[i] = 4
        #
        # -5 + 4 = -1
        # But starting fresh gives:
        # 4
        #
        # So we choose 4.
        current_sum = max(nums[i], current_sum + nums[i])

        # STEP 4:
        # Check whether the current subarray
        # is the best one we have seen.
        #
        # Example:
        # current_sum = 6
        # max_sum = 4
        #
        # 6 is bigger, so update max_sum.
        max_sum = max(max_sum, current_sum)

    # STEP 5:
    # After checking every element,
    # max_sum contains the maximum possible
    # contiguous subarray sum.
    return max_sum
```

```python
def range_sum_query(nums, left, right):

    # STEP 1:
    # Create a prefix sum array.
    #
    # prefix[i] will store the sum of all elements
    # from index 0 up to index i.
    #
    # Example:
    #
    # nums:
    # [2, 4, 1, 5, 3]
    #
    # prefix:
    # [2, 6, 7, 12, 15]
    #
    # Because:
    #
    # prefix[0] = 2
    # prefix[1] = 2 + 4 = 6
    # prefix[2] = 2 + 4 + 1 = 7
    # prefix[3] = 2 + 4 + 1 + 5 = 12
    # prefix[4] = 2 + 4 + 1 + 5 + 3 = 15

    prefix = [0] * len(nums)

    # STEP 2:
    # The first prefix value is simply
    # the first element of nums.
    prefix[0] = nums[0]

    # STEP 3:
    # Build the rest of the prefix array.
    #
    # Current prefix sum =
    # previous prefix sum + current number.
    for i in range(1, len(nums)):
        prefix[i] = prefix[i - 1] + nums[i]

    # STEP 4:
    # Now calculate the range sum.
    #
    # If left == 0:
    # We simply take prefix[right].
    #
    # Example:
    # nums = [2, 4, 1, 5, 3]
    #
    # left = 0
    # right = 2
    #
    # Sum = 2 + 4 + 1 = 7
    #
    # prefix[2] = 7

    if left == 0:
        return prefix[right]

    # STEP 5:
    # If left is not 0, subtract everything
    # before 'left'.
    #
    # Formula:
    #
    # range sum = prefix[right] - prefix[left - 1]
    #
    # Example:
    #
    # nums = [2, 4, 1, 5, 3]
    #
    # left = 1
    # right = 3
    #
    # prefix[3] = 12
    # prefix[0] = 2
    #
    # 12 - 2 = 10
    #
    # Therefore:
    # [4, 1, 5] = 10

    return prefix[right] - prefix[left - 1]
```

```python
def range_sum_query(nums, left, right):
    prefix = [0] * len(nums)
    prefix[0] = nums[0]

    for i in range(1, len(nums)):
        prefix[i] = prefix[i - 1] + nums[i]

    if left == 0:
        return prefix[right]

    return prefix[right] - prefix[left - 1]
```

```python
def build_prefix(nums):

    # STEP 1:
    # Create a prefix array with one extra space.
    #
    # The extra 0 at the beginning makes the
    # range-sum formula easier.
    #
    # Example:
    #
    # nums = [2, 4, 1, 5, 3]
    #
    # prefix = [0, 2, 6, 7, 12, 15]
    #
    # prefix[0] = 0
    # prefix[1] = 2
    # prefix[2] = 2 + 4 = 6
    # prefix[3] = 2 + 4 + 1 = 7
    # ...

    prefix = [0] * (len(nums) + 1)

    # STEP 2:
    # Build the prefix sum.
    #
    # prefix[i + 1] contains the sum of nums
    # from index 0 through index i.
    for i in range(len(nums)):
        prefix[i + 1] = prefix[i] + nums[i]

    return prefix


def range_sum(prefix, left, right):

    # STEP 3:
    # Use the prefix array to calculate the sum
    # from index 'left' to index 'right'.
    #
    # Formula:
    #
    # prefix[right + 1] - prefix[left]
    #
    # Why?
    #
    # prefix[right + 1]
    # = sum from index 0 to right
    #
    # prefix[left]
    # = sum from index 0 to left - 1
    #
    # Subtracting them removes everything before
    # 'left' and leaves only the required range.

    return prefix[right + 1] - prefix[left]
```

```python
def rotate_array(nums, k):

    # STEP 1:
    # Agar array empty hai, toh rotate karne ke liye
    # kuch hai hi nahi.
    if not nums:
        return

    # STEP 2:
    # Agar k array ki length se bada hai,
    # toh k ko directly use karne ki zarurat nahi.
    #
    # Example:
    # nums length = 7
    # k = 10
    #
    # 10 rotations ka effect same hai as:
    # 10 % 7 = 3
    #
    # Isliye:
    k = k % len(nums)

    # STEP 3:
    # Ek helper function banate hain jo
    # array ke left aur right ke beech
    # elements ko reverse karega.
    def reverse(left, right):

        while left < right:

            # Left aur right elements ko swap karo.
            nums[left], nums[right] = nums[right], nums[left]

            # Dono pointers ko andar move karo.
            left += 1
            right -= 1

    # STEP 4:
    # Sabse pehle poore array ko reverse karo.
    #
    # Example:
    #
    # [1, 2, 3, 4, 5, 6, 7]
    #
    # becomes:
    #
    # [7, 6, 5, 4, 3, 2, 1]
    reverse(0, len(nums) - 1)

    # STEP 5:
    # Ab first k elements ko reverse karo.
    #
    # k = 3
    #
    # [7, 6, 5, 4, 3, 2, 1]
    #  └───────┘
    #
    # Reverse:
    #
    # [5, 6, 7, 4, 3, 2, 1]
    reverse(0, k - 1)

    # STEP 6:
    # Ab k ke baad wale remaining elements
    # ko reverse karo.
    #
    # [5, 6, 7, 4, 3, 2, 1]
    #        └────────────┘
    #
    # Reverse:
    #
    # [5, 6, 7, 1, 2, 3, 4]
    reverse(k, len(nums) - 1)

    # STEP 7:
    # Array in-place modify ho chuka hai.
    #
    # Isliye kuch return karne ki zarurat nahi.
```

```python
def rotate_array(nums, k):
    if not nums:
        return

    k %= len(nums)

    def reverse(left, right):
        while left < right:
            nums[left], nums[right] = nums[right], nums[left]
            left += 1
            right -= 1

    reverse(0, len(nums) - 1)
    reverse(0, k - 1)
    reverse(k, len(nums) - 1)
```

```python
if __name__ == "__main__":
    test_cases = [
        ([1, 2, 3, 4, 5, 6, 7], 3),  # Standard case -> [5, 6, 7, 1, 2, 3, 4]
        ([-1, -100, 3, 99], 2),       # Standard case -> [3, 99, -1, -100]
        ([1, 2, 3, 4, 5], 0),         # k is zero (No rotation)
        ([1, 2, 3], 5),               # k larger than length (5 % 3 = 2 steps)
        ([1], 3),                      # Single element
        ([], 2),                       # Empty list
    ]

    for nums, k in test_cases:
        original = nums.copy()
        rotate_array(nums, k)
        print(
            f"Input: {str(original):<25} | k: {k:<2} -> Rotated: {nums}"
        )
```
